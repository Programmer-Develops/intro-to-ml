```python
import pandas as pd;
```

```python
data = {
"Name": ["Ram", "Shyam", "Ghanshyam"],
"Age": [28, 34, 22],
"Salary": [50000, 60000, 45000],
'SP': [2100,3000,5000],
'CP': [2000,2800,5000]
}

df = pd.DataFrame(data)
df
```

```python
df['Total_Cost'] = df['SP']*df['CP']
df
```

```python
df['Diff'] = df['SP']-df['CP']
df
```

```python
df['Prof_Loss'] = df['Diff'].apply(lambda x: 'Profit ' if x>0 else 'Loss')
df
```
